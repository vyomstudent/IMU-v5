### How I currently understand the main components

* **`config/`** → YAML-based configuration for calibration, sensors, gait-event thresholds, MQTT, ZMQ, and rotation matrices.
* **`proto/`** → Protocol Buffer definitions for IMU and pressure data.
* **`libs/calibration_utils.py`** → Mathematical operations used by calibration, such as normalization, cross product, projection, angle calculation, and rotation-matrix construction.
* **`libs/config_reader/`** → Reads YAML configuration files using `yaml.safe_load()`.
* **`libs/mqtt_communication_interface/`** → MQTT connection, subscription, publishing, reconnection, and asyncio integration.  
* **`libs/zmq_communication_interface/`** → Async ZeroMQ PUB/SUB communication with topic-based callbacks.  
* **`data_analysis/`** → Calibration/analysis scripts operating on IMU data.
* **`models/`** → XGBoost models using foot IMU + insole pressure features.
* **`proto/*.proto` + `*_pb2.py`** → Serialization/deserialization layer for IMU and pressure messages.  

**This is my current understanding of the structure.**
